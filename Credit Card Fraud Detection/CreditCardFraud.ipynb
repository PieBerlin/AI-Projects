#### Under sampling

##### build a sample dataset containig simillar distribution of normal transaction and fradulent transactions


```python
#fraudulent transactions  was 492
```

```python
legit_sample = legit.sample(n = 492)
```

## concatinating two dataframes

```python
new_dataset = pd.concat([legit_sample, fraud],axis = 0 )
```

```python
new_dataset.head()
```

```python
new_dataset
```

```python
new_dataset.Class.value_counts()
```

```python
new_dataset.groupby('Class').mean()
```

## splitting dtaset into features an target

```python
Y = new_dataset['Class']
X = new_dataset.drop(columns ='Class',axis =1 )
```

```python
X
```

```python
Y
```

### Split features into training and test data

```python
X_train, X_test , Y_train , Y_test = train_test_split(X, Y,test_size = 0.2, stratify = Y, random_state = 2)
```

## Model training 

```python
model = LogisticRegression()
```

```python
model.fit(X_train, Y_train)
```

### Predicting with test data

```python
training_data_prediction = model.predict(X_train)
training_data_accuracy_score = accuracy_score(training_data_prediction,Y_train)
print(f'Accuracy Score with Training Data is {training_data_accuracy_score:.2f}')
```

```python
### Prediction with test data
```

```python
test_data_prediction = model.predict(X_test)
test_data_accuracy_score = accuracy_score(test_data_prediction,Y_test)
print(f'Accuracy Score with Test Data  is {test_data_accuracy_score:.2f}')
```
